In mathematics, a magic square of order $n$ is composed of $n^2$ strictly positive integers, written in the form of a square array. These numbers are arranged so that their sums over each row, each column, and each main diagonal are equal.

Write a function that takes as input a square (in the form of a list of lists of the same size, where row $i$ represents the $i^{th}$ row of the square) and returns `True` if it is a magic square, and `False` otherwise.


# Magic Square — Mathematical Explanation

A **magic square of order $n$** is an $n \times n$ matrix

$$
M =
\begin{pmatrix}
m_{11} & m_{12} & \cdots & m_{1n} \\
m_{21} & m_{22} & \cdots & m_{2n} \\
\vdots & \vdots & \ddots & \vdots \\
m_{n1} & m_{n2} & \cdots & m_{nn}
\end{pmatrix}
$$

containing $n^2$ strictly positive integers.

The square is **magic** if the sums of:

- every row,
- every column,
- the main diagonal,
- the secondary diagonal,

are all equal to the same value, called the **magic constant**, denoted by $S$.

---

## 1. Row sums

For each row $i$, define

$$
R_i = \sum_{j=1}^{n} m_{ij}.
$$

For the matrix to be magic, all row sums must be equal:

$$
\boxed{
R_1 = R_2 = \cdots = R_n = S
}
$$

or equivalently,

$$
\boxed{
\sum_{j=1}^{n}m_{ij}=S,
\qquad \forall i \in \{1,\ldots,n\}.
}
$$

---

## 2. Column sums

For each column $j$, define

$$
C_j = \sum_{i=1}^{n}m_{ij}.
$$

All columns must also sum to the same magic constant:

$$
\boxed{
C_1=C_2=\cdots=C_n=S
}
$$

or

$$
\boxed{
\sum_{i=1}^{n}m_{ij}=S,
\qquad \forall j \in \{1,\ldots,n\}.
}
$$

---

## 3. Main diagonal

The main diagonal contains

$$
m_{11},m_{22},\ldots,m_{nn}.
$$

Its sum is

$$
D_1=\sum_{i=1}^{n}m_{ii}.
$$

For a magic square,

$$
\boxed{D_1=S}
$$

---

## 4. Secondary diagonal

The secondary diagonal contains

$$
m_{1n},m_{2,n-1},\ldots,m_{n1}.
$$

Its sum is

$$
D_2=\sum_{i=1}^{n}m_{i,n+1-i}.
$$

For a magic square,

$$
\boxed{D_2=S}
$$

---

## 5. Complete condition

A square is therefore magic if

$$
\boxed{
R_1=\cdots=R_n
=
C_1=\cdots=C_n
=
D_1
=
D_2
=
S
}
$$

In other words, **every row, every column, and both main diagonals must have exactly the same sum**.

---

## 6. Example

Consider

$$
M =
\begin{pmatrix}
8 & 1 & 6\\
3 & 5 & 7\\
4 & 9 & 2
\end{pmatrix}.
$$

### Rows

$$
8+1+6=15
$$

$$
3+5+7=15
$$

$$
4+9+2=15
$$

### Columns

$$
8+3+4=15
$$

$$
1+5+9=15
$$

$$
6+7+2=15
$$

### Diagonals

$$
8+5+2=15
$$

and

$$
6+5+4=15.
$$

Therefore,

$$
\boxed{
R_1=R_2=R_3
=C_1=C_2=C_3
=D_1=D_2
=15
}
$$

so this is a **magic square**.

- `option_type` = `'call'` or `'put'`

---

## 1. Black–Scholes–Merton quantities

The model is based on the quantities

$$
d_1
=
\frac{
\ln(S/K)
+
\left(r+\frac{1}{2}\sigma^2\right)T
}{
\sigma\sqrt{T}
}
$$

and

$$
d_2
=
d_1-\sigma\sqrt{T}.
$$

In Python, we need:

```python
import math
```

because:

```python
math.log(x)
```

computes the natural logarithm,

```python
math.sqrt(x)
```

computes the square root, and

```python
math.exp(x)
```

computes the exponential function.

---

## 2. Standard normal distribution

The pricing formulas use the cumulative distribution function of a standard normal random variable:

$$
N(x)=P(Z\leq x),
\qquad
Z\sim\mathcal{N}(0,1).
$$

In Python:

```python
from scipy.stats import norm
```

and:

```python
norm.cdf(x)
```

returns:

$$
N(x).
$$

The Greeks also use the probability density function:

$$
\phi(x)
=
\frac{1}{\sqrt{2\pi}}
e^{-x^2/2}.
$$

In Python:

```python
norm.pdf(x)
```

returns:

$$
\phi(x).
$$

---

The Black–Scholes–Merton price of a European call is:

$$
\boxed{
C
=
SN(d_1)

Ke^{-rT}N(d_2)
}
$$

---

## 4. Put price

The price of a European put is:

$$
\boxed{
P
=
Ke^{-rT}N(-d_2)
-
SN(-d_1)
}
$$

---

## 5. Delta

Delta measures the sensitivity of the option price to the underlying price:

$$
\Delta
=
\frac{\partial V}{\partial S}.
$$

For a call:

$$
\boxed{
\Delta_C=N(d_1)
}
$$

For a put:

$$
\boxed{
\Delta_P=N(d_1)-1
}
$$

---

## 6. Gamma

Gamma measures the sensitivity of Delta to the underlying price:

$$
\Gamma
=
\frac{\partial^2V}{\partial S^2}.
$$

For both calls and puts:

$$
\boxed{
\Gamma
=
\frac{\phi(d_1)}
{S\sigma\sqrt{T}}
}
$$

---

## 7. Vega

Vega measures sensitivity to volatility:

$$
\text{Vega}
=
\frac{\partial V}{\partial \sigma}.
$$

For both calls and puts:

$$
\boxed{
\text{Vega}
=
S\phi(d_1)\sqrt{T}
}
$$

If Vega is expressed for a one-percentage-point move in volatility, we divide by 100:

$$
\boxed{
\text{Vega}_{1\%}
=
\frac{
S\phi(d_1)\sqrt{T}
}{100}
}
$$

---

For a call:

$$
\boxed{
\Theta_C
=
-\frac{S\phi(d_1)\sigma}{2\sqrt{T}}
-
rKe^{-rT}N(d_2)
}
$$

For a put:

$$
\boxed{
\Theta_P
=
-\frac{S\phi(d_1)\sigma}{2\sqrt{T}}
+
rKe^{-rT}N(-d_2)
}
$$

If we want daily Theta, we can divide the annual value by 365.

---

\rho

\frac{\partial V}{\partial r}.

\rho_C

KTe^{-rT}N(d_2)

$$
\boxed{
\rho_P
=
-KTe^{-rT}N(-d_2)
}
$$

If Rho is expressed for a one-percentage-point move in interest rates, we divide by 100.

## Exercise : The Vigenère Cipher

The Vigenère cipher is a substitution encryption system that creates many possible encryptions for the same letter depending on its position in the sentence. This encryption method works using a keyword. Here, we will consider the word **"clef"**. The principle is as follows:

* Transform the keyword into a sequence of integers corresponding to the position of each letter of the word in the alphabet. Here, this gives: *clef* → 3 12 5 6

* To encrypt a sentence, shift: the first letter by 3 letters, the second by 12, the third by 5, the fourth by 6, then repeat: the fifth letter is shifted by 3, the sixth by 12, etc.
